```python
import pandas as pd
hr = pd.read_csv('HR-Employee-Attrition.csv')
attrition_counts = hr['Attrition'].value_counts()
print("--- EMPLOYEE ATTRITION COUNTS ---")
print(attrition_counts)

attrition_percent= hr['Attrition'].value_counts(normalize=True) * 100
print("\n--- ATTRITION PERCENTAGE ---")
print(attrition_percent)
```

```python
import matplotlib.pyplot as plt
income = hr.groupby('Attrition')['MonthlyIncome'].mean()
print("AVERAGE MONTHLY INCOME")
print(income)

income.plot(kind='bar', color=['darkblue', 'red'], figsize =(6,5))
plt.title('Avg Monthly Income vs Attrition')
plt.ylabel('Income ($)')
plt.xticks(rotation=0)
plt.show()
```

```python
overtime = hr.groupby('OverTime')['Attrition'].value_counts(normalize=True).unstack() * 100
overtime.plot(kind='bar', stacked=True, color=['darkblue', 'crimson'], figsize=(7, 5))
plt.title('Attrition Rate by Overtime Work Status')
plt.ylabel('Percentage (%)')
plt.xlabel('Works Overtime?')
plt.legend(title='Left Company?')
plt.xticks(rotation=0)
plt.show()
```
